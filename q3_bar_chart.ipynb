```python
import matplotlib.pyplot as plt
import numpy as np

# Accuracy from benchmark_results.csv (rank = 16)
datasets   = ['GSM8K', 'MATH']
components = ['Top', '25%', '50%', 'Bottom']
accuracy = {
    'Top':    [0.153, 0.038],
    '25%':    [0.126, 0.031],
    '50%':    [0.117, 0.021],
    'Bottom': [0.114, 0.026],
}
colors = ['#93c47d', '#0b4f6c', '#f6b26b', '#3c78d8']   # same colors as the example
```

```python
x = np.arange(len(datasets))   # [0, 1]：兩組的中心位置
width = 0.2                    # 每根柱子的寬度

fig, ax = plt.subplots(figsize=(6, 3.5))
for i, comp in enumerate(components):
    offset = (i - 1.5) * width                 # 讓 4 根柱子以中心點左右排開
    bars = ax.bar(x + offset, accuracy[comp], width, label=comp, color=colors[i])
    ax.bar_label(bars, fmt='%.3f', fontsize=8, padding=2)   # 柱子上方標數字

ax.set_xticks(x, datasets)
ax.set_ylabel('Accuracy')
ax.set_title('Accuracy of different singular-value ranges (rank = 16)')
ax.set_ylim(0, 0.2)
ax.legend(ncol=4, fontsize=9, loc='upper right')
ax.grid(axis='y', alpha=0.4)
ax.set_axisbelow(True)
plt.tight_layout()
plt.savefig('q3_accuracy_bar.png', dpi=200)
plt.show()
```
